```python
import numpy as np
import matplotlib.pyplot as plt

```

```python
data = np.load("joint_log_play.npy")
```

```python
data.shape
```

```python
timesteps, num_envs, num_joints = data.shape
joint_derivatives = np.diff(data[:, :, :].mean(axis=1), axis=0)

plt.figure(figsize=(12, 6))
for joint_idx in range(num_joints):
    if joint_idx in [13,17]:
        plt.plot(data[:, 0, joint_idx] * (180 / np.pi), label=f"Joint {'left' if joint_idx==13 else 'right'}")
plt.xlabel("Timestep")
plt.ylabel("Joint Position")
plt.title("Joint Positions Over Time (First Environment)")
plt.legend(loc="upper right", ncol=2, fontsize="small")
plt.tight_layout()
plt.show()


plt.figure(figsize=(12, 6))
for joint_idx in [13, 17]:
    label = f"Joint {'left' if joint_idx == 13 else 'right'}"
    plt.plot(joint_derivatives[:, joint_idx], label=label)

plt.xlabel("Timestep")
plt.ylabel("Ankle Joint Velocity")
plt.title("Ankle Joint Velocity Over Time (First Environment)")
plt.legend(loc="upper right", ncol=2, fontsize="small")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))
for joint_idx in range(num_joints):
    if joint_idx in [6,11]:
        plt.plot(data[:, :, joint_idx].mean(axis=1) * 1, label=f"Joint {'left' if joint_idx==6 else 'right'}")
plt.xlabel("Timestep")
plt.ylabel("Ankle Joint Position")
plt.title("Ankle Joint Positions Over Time (First Environment)")
plt.legend(loc="upper right", ncol=2, fontsize="small")
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 6))
for joint_idx in [6, 11]:
    label = f"Joint {'left' if joint_idx == 6 else 'right'}"
    plt.plot(joint_derivatives[:, joint_idx], label=label)

plt.xlabel("Timestep")
plt.ylabel("Ankle Joint Velocity")
plt.title("Ankle Joint Velocity Over Time (First Environment)")
plt.legend(loc="upper right", ncol=2, fontsize="small")
plt.tight_layout()
plt.show()
```
